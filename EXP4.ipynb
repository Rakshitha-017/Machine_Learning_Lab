```python
import pandas as pd
```

```python
file_path = "C:/Users/Devops Lab-B2/Desktop/4CB23CB045/forecast.csv"
```

```python
df = pd.read_csv(file_path)
print(df)
```

```python
attribute_value = df.iloc[: , :-1].values
label_value = df.iloc[: , -1].values
num_attributes = len(df.columns)
```

```python
print(num_attributes)
```

```python
print("\n The most general hypothesis : ['?','?','?','?','?','?','?'] \n")
print("\n The most specific hypothesis : ['0','0','0','0','0','0','0']\n")
```

```python
print("\n The initial value of the hypothesis:")
hypothesis = ['0']*(num_attributes - 1)
print(hypothesis)
```

```python
for j, val in enumerate(label_value):
    if val.lower() == 'yes':
        for i in range(len(hypothesis)):
            if hypothesis[i] == '0':
                hypothesis[i] = attribute_value[j, i]
            elif hypothesis[i] != attribute_value[j, i]:
                hypothesis[i] = '?'
```

```python
print("\n The final Hypothesis is : \n")
print(hypothesis)
```
