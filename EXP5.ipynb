```python
import numpy as np 
from sklearn.neighbors import KNeighborsClassifier
```

```python
data = np.random.rand(100)
labels = []
for x in data[:50]:
    if x <= 0.5:
        labels.append(1)
    else:
        labels.append(2)
train_data = data[:50].reshape(-1,1)
train_labels = np.array(labels)
```

```python
test_data = data[50:].reshape(-1,1)
k_value = [1,2,3,4,5,20,30]
for k in k_value:
    knn = KNeighborsClassifier(n_neighbors = k)
    knn.fit(train_data , train_labels)
    predicted = knn.predict(test_data)
    print("\n Results for k =" , k)
    print("Test value\t Predicted class")
    for value , label in zip(test_data.flatten(), predicted):
        print(f"{value:.3f}\t\tClass {label}")
```
