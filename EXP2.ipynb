```python
import pandas as pd
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt 
from sklearn.datasets import fetch_california_housing
```

```python
data = fetch_california_housing(as_frame = True)
housing = data.frame
```

```python
correlation_matrix = housing.corr()
print("Correlation matrix")
print(correlation_matrix)
```

```python
sns.heatmap(correlation_matrix , annot = True , linewidths = 1 , linecolor = 'white'  , cmap ='YlGnBu' )
print('HeatMap')
plt.show()
```

```python
sns.pairplot(housing , height = 5.0, kind = 'scatter' , diag_kind = 'kde' )
print("Paired plot")
plt.show()
```
